```python
import pandas as pd
import math
```

```python

IMPORT_DATASET_PATH = "Iris.csv"
TRAINING_PCT = 0.5
EVALUATING_PCT = 0.25
TESTING_PCT = 0.25

if TRAINING_PCT + EVALUATING_PCT + TESTING_PCT != 1:
    raise Exception("percentages must add to 100!")
```

```python
import_data = pd.read_csv(IMPORT_DATASET_PATH)
IMPORT_DATA_LENGTH = len(import_data)

TRAINING_NUM = math.floor(TRAINING_PCT * IMPORT_DATA_LENGTH)
EVALUATING_NUM = math.floor(EVALUATING_PCT * IMPORT_DATA_LENGTH)
TESTING_NUM = IMPORT_DATA_LENGTH - TRAINING_NUM - EVALUATING_NUM

training_data = import_data.sample()
import_data = import_data.drop(training_data.index)

evaluating_data = import_data.sample(n=EVALUATING_NUM, random_state=42)
import_data = import_data.drop(evaluating_data.index)

testing_data = import_data
```

```python
print(training_data)
print(testing_data)
print(evaluating_data)
```

```python
training_data.to_csv(f"{IMPORT_DATASET_PATH.replace(".csv", "_")}training_data.csv")
evaluating_data.to_csv(f"{IMPORT_DATASET_PATH.replace(".csv", "_")}evaluating_data.csv")
testing_data.to_csv(f"{IMPORT_DATASET_PATH.replace(".csv", "_")}testing_data.csv")
```
